```python
# ==============================================================================
# STEP 0: SETUP AND INSTALLATIONS
# ==============================================================================
print("--- Step 0: Installing MNE and importing libraries ---")
!pip install mne -q

import os
import gc
import mne
import numpy as np
from google.colab import drive
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, cohen_kappa_score
from sklearn.utils import class_weight

import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv1D, MaxPooling1D, LSTM, Dense, Dropout, BatchNormalization
from tensorflow.keras.layers import Attention, Concatenate, GlobalAveragePooling1D, Multiply, Reshape
from tensorflow.keras.utils import to_categorical
# ---> ADDED: Callbacks for preventing overfitting <---
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

print("Setup complete. Libraries imported successfully. ✅")

# ==============================================================================
# STEP 1: MOUNT DRIVE AND DISCOVER DATA FILES
# ==============================================================================
print("\n--- Step 1: Mounting Google Drive and finding data files ---")
drive.mount('/content/drive')

base_path = "/content/drive/MyDrive/sleep-data/"

try:
    all_files_in_dir = os.listdir(base_path)
    psg_files = sorted([f for f in all_files_in_dir if 'PSG.edf' in f])
    hyp_files = sorted([f for f in all_files_in_dir if 'Hypnogram.edf' in f])

    if not psg_files:
        raise FileNotFoundError("No PSG files found in the specified directory.")

    print(f"Found {len(psg_files)} PSG files and {len(hyp_files)} Hypnogram files.")

except FileNotFoundError as e:
    print(f"Error: Could not find the directory '{base_path}'. Please check the path.")
    exit()

# ==============================================================================
# STEP 2: PROCESS EACH SUBJECT (MEMORY OPTIMIZED)
# ==============================================================================
print("\n--- Step 2: Processing subjects and preparing raw epoch data ---")

all_epochs_data = []
all_labels = []
groups = []

for subject_index, psg_filename in enumerate(psg_files):
    subject_base_id = psg_filename.split('E')[0]
    print(f"\nProcessing Subject {subject_index + 1}/{len(psg_files)}: {subject_base_id}")

    matching_hyp_file = next((h for h in hyp_files if h.startswith(subject_base_id)), None)

    if not matching_hyp_file:
        print(f"  !!! Warning: Could not find a hypnogram for {psg_filename}. Skipping.")
        continue

    full_psg_path = os.path.join(base_path, psg_filename)
    full_hyp_path = os.path.join(base_path, matching_hyp_file)

    try:
        raw = mne.io.read_raw_edf(full_psg_path, preload=True, stim_channel='auto', verbose='error')
        annot = mne.read_annotations(full_hyp_path)
        raw.set_annotations(annot, verbose=False)
        raw.pick_channels(['EEG Fpz-Cz', 'EOG horizontal', 'EMG submental'])
        raw.filter(l_freq=0.3, h_freq=35, verbose=False)
        raw.resample(100, verbose=False) # Downsample to 100 Hz

        annotation_desc_2_event_id = {'Sleep stage W': 0, 'Sleep stage 1': 1, 'Sleep stage 2': 2, 'Sleep stage 3': 3, 'Sleep stage 4': 3, 'Sleep stage R': 4}
        events, _ = mne.events_from_annotations(raw, event_id=annotation_desc_2_event_id, chunk_duration=30.)
        epochs = mne.Epochs(raw=raw, events=events, tmin=0, tmax=30, baseline=None, preload=True, verbose=False)

        # 1. IMMEDIATE DOWNCAST TO FLOAT32 (Saves 50% RAM)
        data = epochs.get_data().astype(np.float32)

        # 2. PER-SUBJECT NORMALIZATION
        for c in range(data.shape[1]):
            channel_data = data[:, c, :]
            mean = np.mean(channel_data)
            std = np.std(channel_data)
            if std > 0:
                data[:, c, :] = (channel_data - mean) / std

        all_epochs_data.append(data)
        all_labels.extend(epochs.events[:, -1])
        groups.extend([subject_index] * len(epochs))

        # Clean up MNE objects immediately to free RAM
        del raw
        del epochs
        gc.collect()

        print(f"  Successfully processed. Found {data.shape[0]} epochs.")

    except Exception as e:
        print(f"  !!! Error processing {psg_filename}: {e}")

X = np.concatenate(all_epochs_data, axis=0)
y = np.array(all_labels)
groups = np.array(groups)
y_one_hot = to_categorical(y)
num_classes = y_one_hot.shape[1]

# Delete the massive list of arrays to free up RAM
del all_epochs_data
gc.collect()

# ==============================================================================
# STEP 3: DATA SPLITTING AND MODALITY SEPARATION
# ==============================================================================
print("\n--- Step 3: Splitting Data ---")

# Transpose to the shape the model expects: (epochs, timepoints, channels)
X = np.transpose(X, (0, 2, 1))

# Split based on subjects to prevent data leakage
gss = GroupShuffleSplit(n_splits=1, test_size=0.3, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups))

X_train, X_test = X[train_idx], X[test_idx]
y_train_one_hot, y_test_one_hot = y_one_hot[train_idx], y_one_hot[test_idx]
y_test_labels = y[test_idx]

# Delete the massive combined 'X' array now that it is split!
del X
gc.collect()

# SPLIT INTO MODALITIES FOR MULTI-BRANCH FUSION
X_train_eeg = X_train[:, :, 0:1]
X_train_eog = X_train[:, :, 1:2]
X_train_emg = X_train[:, :, 2:3]

X_test_eeg = X_test[:, :, 0:1]
X_test_eog = X_test[:, :, 1:2]
X_test_emg = X_test[:, :, 2:3]

print(f"Data split. EEG training shape: {X_train_eeg.shape}")

# ==============================================================================
# STEP 4: BUILD MULTIMODAL ATTENTION NETWORK
# ==============================================================================
print("\n--- Step 4: Building Multimodal Attention Model ---")

def channel_attention_module(x, ratio=8):
    """Squeeze-and-Excitation block for Channel Attention."""
    channels = x.shape[-1]
    squeeze = GlobalAveragePooling1D()(x)
    excitation = Dense(channels // ratio, activation='relu')(squeeze)
    excitation = Dense(channels, activation='sigmoid')(excitation)
    excitation = Reshape((1, channels))(excitation)
    return Multiply()([x, excitation])

def build_specialist_branch(input_layer, name_prefix):
    """Extracts features for a specific physiological signal."""
    # CNN Block 1 + Channel Attention
    x = Conv1D(filters=64, kernel_size=50, activation='relu', padding='same')(input_layer)
    x = BatchNormalization()(x)
    x = channel_attention_module(x)
    x = MaxPooling1D(pool_size=8)(x)
    x = Dropout(0.5)(x)

    # CNN Block 2 + Channel Attention
    x = Conv1D(filters=128, kernel_size=8, activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = channel_attention_module(x)
    x = MaxPooling1D(pool_size=4)(x)
    x = Dropout(0.5)(x)

    # LSTM for temporal dynamics
    x = LSTM(units=64, return_sequences=True, name=f'{name_prefix}_lstm')(x)
    return x

# Define the 3 distinct inputs
input_eeg = Input(shape=(X_train_eeg.shape[1], 1), name='EEG_Input')
input_eog = Input(shape=(X_train_eog.shape[1], 1), name='EOG_Input')
input_emg = Input(shape=(X_train_emg.shape[1], 1), name='EMG_Input')

# Extract features independently
lstm_eeg = build_specialist_branch(input_eeg, 'EEG')
lstm_eog = build_specialist_branch(input_eog, 'EOG')
lstm_emg = build_specialist_branch(input_emg, 'EMG')

# --- THE FUSION ENGINE (ATTENTION) ---
# 1. EEG queries itself to solidify context (Self-Attention)
attn_self_eeg = Attention(use_scale=True, name='Self_Attn_EEG')([lstm_eeg, lstm_eeg])

# 2. EEG queries EOG for eye movement confirmation (Cross-Attention)
attn_eeg_eog = Attention(use_scale=True, name='Cross_Attn_EEG_EOG')([lstm_eeg, lstm_eog])

# 3. EEG queries EMG for muscle tone confirmation (Cross-Attention)
attn_eeg_emg = Attention(use_scale=True, name='Cross_Attn_EEG_EMG')([lstm_eeg, lstm_emg])

# Combine the attended features
merged_features = Concatenate(axis=-1)([attn_self_eeg, attn_eeg_eog, attn_eeg_emg])

# Collapse the time dimension and classify
pooled_features = GlobalAveragePooling1D()(merged_features)
dense_out = Dropout(0.5)(pooled_features)
dense_out = Dense(64, activation='relu')(dense_out)
output_layer = Dense(num_classes, activation='softmax')(dense_out)

model = Model(inputs=[input_eeg, input_eog, input_emg], outputs=output_layer)
optimizer = tf.keras.optimizers.Adam(learning_rate=0.0005)
model.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy'])

model.summary()

# Calculate class weights for imbalance
class_weights = class_weight.compute_class_weight('balanced', classes=np.unique(y), y=y[train_idx])
class_weights_dict = dict(enumerate(class_weights))

# ---> ANTI-OVERFITTING CALLBACKS ADDED HERE <---
# 1. Early Stopping: Stop if val_loss doesn't improve for 12 epochs
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=12,
    restore_best_weights=True,
    verbose=1
)

# 2. Reduce LR: Cut learning rate in half if val_loss stalls for 5 epochs
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=5,
    min_lr=0.00001,
    verbose=1
)

# Train the model with the callbacks
history = model.fit(
    [X_train_eeg, X_train_eog, X_train_emg], y_train_one_hot,
    epochs=75,
    batch_size=64,
    validation_data=([X_test_eeg, X_test_eog, X_test_emg], y_test_one_hot),
    class_weight=class_weights_dict,
    callbacks=[early_stopping, reduce_lr] # Inject callbacks here
)

print("Model training complete! 🚀")

# ==============================================================================
# STEP 5: EVALUATION
# ==============================================================================
print("\n--- Step 5: Evaluating the model's performance ---")

# Plot training history
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Model Accuracy')
plt.legend()
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.legend()
plt.show()

# Predict using the 3 test branches
y_pred_probs = model.predict([X_test_eeg, X_test_eog, X_test_emg])
y_pred = np.argmax(y_pred_probs, axis=1)

accuracy = accuracy_score(y_test_labels, y_pred)
kappa = cohen_kappa_score(y_test_labels, y_pred)

print(f"\nModel Accuracy on Unseen Subjects: {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Cohen's Kappa Score: {kappa:.4f}")

print("\n--- Classification Report ---")
target_names = ['Wake', 'N1', 'N2', 'N3/N4', 'REM']
print(classification_report(y_test_labels, y_pred, target_names=target_names, zero_division=0))

print("\n--- Confusion Matrix ---")
cm = confusion_matrix(y_test_labels, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=target_names, yticklabels=target_names)
plt.title('Confusion Matrix on Unseen Subjects')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.show()
```
